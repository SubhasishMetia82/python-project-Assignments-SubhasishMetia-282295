```python
#1
l=[]
for i in range(10):
  n=int(input("Enter number:"))
  l.append(n)
print("New List:",l)

max=0
min=l[0]
for i in range(len(l)):
  if (l[i]>max):
    max=l[i]
  if (l[i]<min):
    min=l[i]

print("Maximum number:",max)
print("Minimum number:",min)
```

```python
#2
l=[]
num=int(input("Enter range:"))
for i in range(num):
  n=int(input("Enter number:"))
  l.append(n)
print("List:",l)

New_List=[]
c=0
for i in range(len(l)):
  for j in range(i):
    if(l[i]==l[j]):
      c+=1
  if (c==0):
    New_List.append(l[i])
  c=0

print("New List:",New_List)
```

```python
#3
l=[]
num=int(input("Enter range:"))
for i in range(num):
  n=int(input("Enter number:"))
  l.append(n)
print("List:",l)

l1=[]
l2=[]
for i in range(len(l)):
  if (l[i]%2==0):
     l1.append(l[i])
  if (l[i]%2!=0):
     l2.append(l[i])

print("Even list:",l1)
print("Odd list:",l2)
```
