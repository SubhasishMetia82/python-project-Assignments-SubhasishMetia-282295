```python
#2
def celcius(x):
  f=32+(1.8*x)
  print("Celcius to Fahrenheit:",f)

def fahrenheit(x):
  c=(x-32)*(5/9)
  print("Fahrenheit to Celcius:",c)

def kelvin(x):
  k=x+273.15
  print("Celcius to Kelvin:",k)

x=int(input("Enter temperature:"))
celcius(x)
fahrenheit(x)
kelvin(x)
```

```python
#3
def module_factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers"
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

```python
#3(continue)
import module_factorial

num = int(input("Enter a number: "))
print("Factorial of", num, "is", module_factorial.factorial(num))
```

```python
#4
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

num = int(input("Enter a number: "))
if is_prime(num):
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```
