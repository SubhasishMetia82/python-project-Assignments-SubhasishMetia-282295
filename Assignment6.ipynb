```python
#1(Tuple)
t = (1, 2, 2, 3, 1, 4, 3)
unique = []
for x in t:
    if x not in unique:
        unique.append(x)
print("T1:", tuple(unique))
```

```python
#2
students = (("Riya", 85), ("Aman", 92), ("Soham", 78), ("Priya", 89))
result = sorted(students, key=lambda s: s[1], reverse=True)
for name, marks in result:
    print("T2:", name, marks)
```

```python
#3
nums = []
for i in range(5):
    nums.append(int(input("Enter number: ")))
nums = tuple(nums)
total = sum(nums)
print("T3: Sum =", total, " Average =", total / 5)
```

```python
#4
t = (10, 20, 10, 30, 10, 40, 20)
print("T4: 10 occurs", t.count(10), "times")
```

```python
#5
t = (4, 9, 2, 15, 7)
big = t[0]
small = t[0]
for x in t:
    if x > big:
        big = x
    if x < small:
        small = x
print("T5: Largest =", big, " Smallest =", small)
```

```python
#6(Sets)
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print("S1: Union =", A | B)
print("S1: Intersection =", A & B)
print("S1: A - B =", A - B)
```

```python
#8
s1 = {1, 2}
s2 = {1, 2, 3, 4}
print("S3: s1 is subset of s2?", s1.issubset(s2))
print("S3: s2 is superset of s1?", s2.issuperset(s1))
```

```python
#10(Dictionary)
marks = {"Riya": 85, "Aman": 92, "Soham": 78, "Priya": 89, "Rahul": 70}
for name in marks:
    print("D1:", name, "->", marks[name])
```

```python
#11
sentence = input("Enter a sentence: ")   # e.g. python is easy and python is useful
freq = {}
for word in sentence.split():
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1
print("D2:", freq)
```

```python
#12
d1 = {"Pen": 10, "Pencil": 5, "Book": 3}
d2 = {"Pen": 4, "Pencil": 7, "Eraser": 6}
merged = d1.copy()
for item in d2:
    if item in merged:
        merged[item] += d2[item]
    else:
        merged[item] = d2[item]
print("D3:", merged)
```
