```python
#1
n=int(input("Enter number:"))
num=n
s=0
while(n>0):
  d=n%10
  s+=d
  n//=10
print("Sum of the digits of ",num," is ",s)
```

```python
#2
n=int(input("Enter number:"))
num=n
s=0
while(n>0): #when required use n>=0,otherwise not
  d=n%10
  s=s*10+d
  n//=10
print("Reverse of ",num," is ",s)
```

```python
#3
s=0
c=1
while(c>0):
  n=int(input("Enter number:"))
  if(n==0):
    break
  s+=n
  c+=1
print("Sum of the number inputed",s)
```
